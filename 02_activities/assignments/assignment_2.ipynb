# Preprocessing

Create two [Column Transformers](https://scikit-learn.org/stable/modules/generated/sklearn.compose.ColumnTransformer.html), called preproc1 and preproc2, with the following guidelines:

- Numerical variables

    * (Preproc 1 and 2) Scaling: use a scaling method of your choice (Standard, Robust, Min-Max). 
    * Preproc 2 only: 
        
        + Choose a transformation for any of your input variables (or several of them). Evaluate if this transformation is convenient.

- Categorical variables: 
    
    * (Preproc 1 and 2) Apply [one-hot encoding](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.OneHotEncoder.html) where appropriate.


+ The only difference between preproc1 and preproc2 is the non-linear transformation of the numerical variables.
    


### Preproc 1

Create preproc1 below.

+ Numeric: scaled variables, no other transforms.
+ Categorical: one-hot encoding.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_features = ['coord_x', 'coord_y', 'ffmc', 'dmc', 'dc', 'isi', 'temp', 'rh', 'wind', 'rain']
categorical_features = ['month', 'day']
preproc1 = ColumnTransformer([
    ('num', StandardScaler(), numeric_features),
    ('cat', OneHotEncoder(drop='first', handle_unknown='ignore'), categorical_features)
])
preproc1
```

### Preproc 2

Create preproc1 below.

+ Numeric: scaled variables, non-linear transformation to one or more variables.
+ Categorical: one-hot encoding.

```python
X[['dmc', 'rain']].describe()
```

```python
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer
other_features = [col for col in numeric_features if col != 'rain']

rain_transformer = Pipeline([
    ('rain_log', FunctionTransformer(np.log1p, validate=False)),
    ('rain_scaler', StandardScaler())
])

preproc2 = ColumnTransformer([
    ('num', StandardScaler(), other_features),
    ('rain', rain_transformer, ['rain']),
    ('cat', OneHotEncoder(drop='first', handle_unknown='ignore'), categorical_features)
])
preproc2
```

```python
ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), other_features),
        ('rain', Pipeline(steps=[
            ('rain_log', FunctionTransformer(func=np.log1p)), ('scale', StandardScaler())]),
        ['rain']),
        ('cat', OneHotEncoder(handle_unknown='ignore'), ['month', 'day'])])
```

Create a [model pipeline](https://scikit-learn.org/stable/modules/generated/sklearn.pipeline.Pipeline.html):

## Regressor

+ Use a regression model to perform a prediction. 

    - Choose a baseline regressor, tune it (if necessary) using grid search, and evaluate it using cross-validation.
    - Choose a more advance regressor, tune it (if necessary) using grid search, and evaluate it using cross-validation.
    - Both model choices are up to you, feel free to experiment.

```python
# Pipeline A = preproc1 + baseline
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge
pp_A = Pipeline([
    ('preprocessing', preproc1),
    ('regressor', Ridge(alpha=1.0))
])
pp_A 
```

```python
# Pipeline B = preproc2 + baseline
pp_B = Pipeline([
    ('preprocessing', preproc2),
    ('regressor', Ridge(alpha=1.0))
])  
pp_B
```

```python
# Pipeline C = preproc1 + advanced model
from sklearn.ensemble import RandomForestRegressor
pp_C = Pipeline([
    ('preprocessing', preproc1),
    ('regressor', RandomForestRegressor(random_state=42))
])
pp_C
```

```python
# Pipeline D = preproc2 + advanced model
pp_D = Pipeline([
    ('preprocessing', preproc2),
    ('regressor', RandomForestRegressor(random_state=42))
])
pp_D

    
```

# Tune Hyperparams

+ Perform GridSearch on each of the four pipelines. 
+ Tune at least one hyperparameter per pipeline.
+ Experiment with at least four value combinations per pipeline.

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
print("Training:", X_train.shape)
print("Testing:", X_test.shape)
```

```python
from sklearn.model_selection import GridSearchCV
par_grid_a = {
    'regressor__alpha': [0.1, 1, 10, 100]
}

grid_search_a = GridSearchCV(pp_A, param_grid=par_grid_a, cv=5, scoring='neg_mean_absolute_error', error_score='raise')

grid_search_a.fit(X_train, Y_train)

print("Best parameters for Pipeline A:", grid_search_a.best_params_)
print("Best CV MAE:", -grid_search_a.best_score_)
```

```python
par_grid_b = {
    'regressor__alpha': [0.1, 1, 10, 100]}

grid_search_b = GridSearchCV(pp_B, param_grid=par_grid_b, cv=5, scoring='neg_mean_absolute_error', error_score='raise')

grid_search_b.fit(X_train, Y_train)

print("Best parameters for Pipeline B:", grid_search_b.best_params_)
print("Best CV MAE:", -grid_search_b.best_score_)
```

```python
par_grid_c = {
    'regressor__n_estimators': [20, 50, 100, 150]}

grid_search_c = GridSearchCV(pp_C, param_grid=par_grid_c, cv=5, scoring='neg_mean_absolute_error', error_score='raise')

grid_search_c.fit(X_train, Y_train)

print("Best parameters for Pipeline C:", grid_search_c.best_params_)
print("Best CV MAE:", -grid_search_c.best_score_)
```

```python
par_grid_d = {
    'regressor__n_estimators': [20, 50, 100, 150]}
grid_search_d = GridSearchCV(pp_D, param_grid=par_grid_d, cv=5, scoring='neg_mean_absolute_error', error_score='raise')
grid_search_d.fit(X_train, Y_train)
print("Best parameters for Pipeline D:", grid_search_d.best_params_)
print("Best CV MAE:", -grid_search_d.best_score_)
```

# Evaluate

+ Which model has the best performance?
The pipeline with the lowest MAE was selected as the best performing model.

```python
import pandas as pd

results = pd.DataFrame({
    'Pipeline': ['A', 'B', 'C', 'D'],
    'CV MAE': [
        -grid_search_a.best_score_,
        -grid_search_b.best_score_,
        -grid_search_c.best_score_,
        -grid_search_d.best_score_
    ]
})

results = results.sort_values('CV MAE')
print(results)
```

# Export

+ Save the best performing model to a pickle file.

```python
from sklearn.metrics import mean_absolute_error
best_pipeline = grid_search_b.best_estimator_
predictions = best_pipeline.predict(X_test)
test_mae = mean_absolute_error(Y_test, predictions)

print("Best model: Pipeline B")
print("Test MAE for the best pipeline:", test_mae)
```
